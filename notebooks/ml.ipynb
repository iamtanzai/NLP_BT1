# GIAI ĐOẠN 5: HUẤN LUYỆN MÔ HÌNH HỌC MÁY (MACHINE LEARNING)

## 1. Nhiệm vụ của thành viên phụ trách Machine Learning
- **Mục tiêu**: Sử dụng các đặc trưng đã sinh ra từ Giai đoạn 4 tại `models/features/` để huấn luyện 2 thuật toán trên cả 3 loại đặc trưng.
- **2 thuật toán học máy**:
  1. **Naive Bayes** (`MultinomialNB` cho BoW/TF-IDF, `GaussianNB` cho Word2Vec).
  2. **Support Vector Machine** (`LinearSVC`).
- **3 loại đặc trưng**:
  1. **Bag of Words (BoW)**
  2. **TF-IDF**
  3. **Word2Vec**
- **Tổng cộng huấn luyện 6 mô hình** và lưu vào thư mục: `models/classifiers/`
  - `model_nb_bow.pkl` & `model_svm_bow.pkl`
  - `model_nb_tfidf.pkl` & `model_svm_tfidf.pkl`
  - `model_nb_word2vec.pkl` & `model_svm_word2vec.pkl`

### Bước 1: Khai báo thư viện và nạp dữ liệu đặc trưng từ `models/features/`

```python
import joblib
import numpy as np
import pandas as pd
from pathlib import Path

# Khai báo các mô hình học máy: Naive Bayes và SVM
from sklearn.naive_bayes import MultinomialNB, GaussianNB
from sklearn.svm import LinearSVC

# Thiết lập thư mục nạp đặc trưng và thư mục lưu 6 mô hình
MODEL_FEATURE_DIR = Path("../models/features")
MODEL_TRAINED_DIR = Path("../models/classifiers")
MODEL_TRAINED_DIR.mkdir(parents=True, exist_ok=True)

data_path = MODEL_FEATURE_DIR / "feature_data.joblib"

# Kiểm tra tệp đặc trưng
if not data_path.exists():
    raise FileNotFoundError(f"Không tìm thấy tệp đặc trưng tại {data_path}. Hãy chạy notebooks/feature.ipynb trước!")

# Nạp toàn bộ dữ liệu đặc trưng
feature_bundle = joblib.load(data_path)

# 1. Tập nhãn
y_train = feature_bundle["y_train"]
y_val   = feature_bundle["y_val"]
y_test  = feature_bundle["y_test"]

# 2. Ma trận đặc trưng BoW (Bag of Words)
X_train_bow = feature_bundle["bow"]["X_train"]
X_val_bow   = feature_bundle["bow"]["X_val"]
X_test_bow  = feature_bundle["bow"]["X_test"]

# 3. Ma trận đặc trưng TF-IDF
X_train_tfidf = feature_bundle["tfidf"]["X_train"]
X_val_tfidf   = feature_bundle["tfidf"]["X_val"]
X_test_tfidf  = feature_bundle["tfidf"]["X_test"]

# 4. Ma trận đặc trưng Word2Vec
X_train_w2v = feature_bundle["word2vec"]["X_train"]
X_val_w2v   = feature_bundle["word2vec"]["X_val"]
X_test_w2v  = feature_bundle["word2vec"]["X_test"]

print("✅ NẠP DỮ LIỆU ĐẶC TRƯNG THÀNH CÔNG:")
print(f"- Nhãn huấn luyện (y_train) : {len(y_train)} mẫu")
print(f"- 1. Đặc trưng BoW shape    : Train={X_train_bow.shape}, Val={X_val_bow.shape}, Test={X_test_bow.shape}")
print(f"- 2. Đặc trưng TF-IDF shape : Train={X_train_tfidf.shape}, Val={X_val_tfidf.shape}, Test={X_test_tfidf.shape}")
print(f"- 3. Đặc trưng Word2Vec shape: Train={X_train_w2v.shape}, Val={X_val_w2v.shape}, Test={X_test_w2v.shape}")
```

### Bước 2: Huấn luyện 6 mô hình (2 thuật toán × 3 loại đặc trưng)
- **Nhóm 1**: Huấn luyện với đặc trưng **Bag of Words (BoW)** -> `model_nb_bow`, `model_svm_bow`.
- **Nhóm 2**: Huấn luyện với đặc trưng **TF-IDF** -> `model_nb_tfidf`, `model_svm_tfidf`.
- **Nhóm 3**: Huấn luyện với đặc trưng **Word2Vec** -> `model_nb_word2vec`, `model_svm_word2vec`.
- Toàn bộ 6 mô hình sau khi `fit(X_train, y_train)` sẽ được lưu vào `models/classifiers/`.

```python
print("="*65)
print("TIẾN HÀNH HUẤN LUYỆN 6 MÔ HÌNH HỌC MÁY:")
print("="*65)

# ==============================================================================
# 1. HUẤN LUYỆN VỚI ĐẶC TRƯNG BAG OF WORDS (BoW)
# ==============================================================================
print("--> [1/3] Đang huấn luyện mô hình với đặc trưng BoW...")
model_nb_bow = MultinomialNB()
model_nb_bow.fit(X_train_bow, y_train)

model_svm_bow = LinearSVC(random_state=42)
model_svm_bow.fit(X_train_bow, y_train)

# ==============================================================================
# 2. HUẤN LUYỆN VỚI ĐẶC TRƯNG TF-IDF
# ==============================================================================
print("--> [2/3] Đang huấn luyện mô hình với đặc trưng TF-IDF...")
model_nb_tfidf = MultinomialNB()
model_nb_tfidf.fit(X_train_tfidf, y_train)

model_svm_tfidf = LinearSVC(random_state=42)
model_svm_tfidf.fit(X_train_tfidf, y_train)

# ==============================================================================
# 3. HUẤN LUYỆN VỚI ĐẶC TRƯNG WORD2VEC
# Lưu ý: Vector Word2Vec chứa số thực liên tục (có giá trị âm) nên Naive Bayes dùng GaussianNB
# ==============================================================================
print("--> [3/3] Đang huấn luyện mô hình với đặc trưng Word2Vec...")
model_nb_w2v = GaussianNB()
model_nb_w2v.fit(X_train_w2v, y_train)

model_svm_w2v = LinearSVC(random_state=42)
model_svm_w2v.fit(X_train_w2v, y_train)

print("\n✅ ĐÃ HUẤN LUYỆN THÀNH CÔNG TẤT CẢ 6 MÔ HÌNH!")

# ==============================================================================
# 4. LƯU CẢ 6 MÔ HÌNH VÀO THƯ MỤC models/classifiers/
# ==============================================================================
# Lưu mô hình BoW
joblib.dump(model_nb_bow,  MODEL_TRAINED_DIR / "model_nb_bow.pkl")
joblib.dump(model_svm_bow, MODEL_TRAINED_DIR / "model_svm_bow.pkl")

# Lưu mô hình TF-IDF
joblib.dump(model_nb_tfidf,  MODEL_TRAINED_DIR / "model_nb_tfidf.pkl")
joblib.dump(model_svm_tfidf, MODEL_TRAINED_DIR / "model_svm_tfidf.pkl")

# Lưu mô hình Word2Vec
joblib.dump(model_nb_w2v,  MODEL_TRAINED_DIR / "model_nb_word2vec.pkl")
joblib.dump(model_svm_w2v, MODEL_TRAINED_DIR / "model_svm_word2vec.pkl")

print("\nĐã lưu 6 mô hình vào thư mục: models/classifiers/")
print("1. models/classifiers/model_nb_bow.pkl")
print("2. models/classifiers/model_svm_bow.pkl")
print("3. models/classifiers/model_nb_tfidf.pkl")
print("4. models/classifiers/model_svm_tfidf.pkl")
print("5. models/classifiers/model_nb_word2vec.pkl")
print("6. models/classifiers/model_svm_word2vec.pkl")
```
